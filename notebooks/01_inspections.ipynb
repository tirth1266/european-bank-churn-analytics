```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt

pd.set_option("display.max_columns",None)

DATA_PATH = "../data/raw/European_Bank.csv"
bank = pd.read_csv(DATA_PATH)
```

```python
print(bank.shape)
```

```python
# to check data integrity.
print("Duplicate rows:",bank.duplicated().sum())
print("CustomerId unique:",bank["CustomerId"].is_unique)
print(bank["Year"].value_counts())
```

```python
bank.describe().T
```

```python
# to check every value 
for c in ["Geography","Gender","NumOfProducts","HasCrCard","IsActiveMember"]:
    print(bank[c].value_counts(),"\n")
print("Zero balance share:",(bank["Balance"]==0).mean())
print("Unique Surnames:", bank["Surname"].unique())
```
